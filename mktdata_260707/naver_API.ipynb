```python
# range 몇 번 반복할거니?
# index변수에는 0~9까지 저장됨
#1~100
#101~200
# ...
# 901~1000
# 찾아오고 싶어

start = 1
num = 0

client_id = "25nggv6gop"
client_secret = "bK84gLRILdIrPOz40Z6DIw4hQxVbbd9jgpvQXBqd"

excel_file = openpyxl.Workbook()
# 엑티브 해줘야 함
excel_sheet = excel_file.active
excel_sheet.column_dimensions["B"].width = 100
excel_sheet.column_dimensions["C"].width = 100
excel_sheet.append(["No", "제목", "링크"])

for index in range(10):
    start_number = start + (index * 100)

    naver_open_api = f"https://naverapihub.apigw.ntruss.com/search/v1/news?query=나이키&display=100&start={str(start_number)}&sort=date&format=json"
    header_params = {
        "X-NCP-APIGW-API-KEY-ID": client_id,
        "X-NCP-APIGW-API-KEY": client_secret
    }        
    
    res = requests.get(naver_open_api, headers=header_params)

    if res.status_code == 200 :
        data =res.json()
        for item in data["items"]:
            num += 1
            excel_sheet.append([num, item["title"], item["link"]])          
    else :
        print("Error Code:", res.status_code)

excel_file.save("naver_api_result.xlsx")
excel_file.close()

print("naver_api_result.xlsx 파일 저장 완료!😀")
```
