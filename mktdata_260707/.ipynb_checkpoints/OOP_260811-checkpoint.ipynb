```python
# 컴퓨터 공학과 1학년 1학기 : C언어 > PP OOP를 배움, 이유없이 그냥 가르치진 않을거야잉?
# PP : Procedural Programming = 절차지향 프로그래밍(1950~1990)(기계중심)
# C언어, 포트란, 파스칼, 코볼

# OOP : Object Oriented Programming = 객체지향 프로그래밍(1990~현재)(인간중심)
# 파이썬, 자바스크립트, 자바

# 냉장고에 코끼리를 넣자!!!
# 절차지향 방식으로 코드를 작성

# 냉장고 문을 연다
# 코끼리를 냉장고에 넣는다
# 냉장고를 닫는다

# 냉장고에 넣어야 하는 코끼리 100마리
# 냉장고 문을 100번 열어야 함
# 냉장고에 100마리 코끼리 개별적으로 넣어야 함
# 냉장고 문을 100번 닫아야 함

# 절차지향 프로그래밍 언어 > 코드가 엄청 길다 = 알아야할 문법 많음

# 객체지향 방식으로 코끼리를 냉장고에 넣는다는 코드를 작성?!
# 객체지향은 결국 절차지향이 가질 수 밖에 없는 필연적인 단점, 한계 극복

# 앞으로 코끼리처럼 생긴 동물이 오면 그냥 알아서 코끼리라고 인식을 하고 준비해놓은 기능들을 통해서 보다 수월하게
# 문도 열고, 코끼리도 넣고, 문도 닫도록 하고 싶음!!

# 코끼리라는 동물의 정보 및 형태에 대한 설계도를 작성
# 코끼리라는 설계도에 근거해서 코끼리1이라는 객체를 하나 생성
# 냉장고라는 사물의 정보 및 형태에 대한 설계도를 작성
# 냉장고라는 설계도에 근거해서 냉장고1이라는 객체를 하나 생성
# 코끼리1이라는 객체에 사전에 학습시켜놓은 기능을 호출해서 냉장고1의 문을 열게끔 사전에 세팅
# 코끼리1이라는 객체에 사전에 학습시켜놓은 기능을 호출해서 냉장고1의 안으로 들어가게끔 사전에 세팅
# 코끼리1이라는 객체에 사전에 학습시켜놓은 기능을 호출해서 냉장고1의 문을 닫게끔 사전에 세팅

# 코끼리 100마리 => 코끼리라는 설계도에서부터 100마리는 다 탄생 => 내가 냉장고에 어떻게, 냉장고가 어떻게 생겼는지도!

# 절차지향 언어는 유사한 기능 구현임에도 불구하고, 매번 절차에 맞춰서 문법에 따라 코드를 작썽 불편
# 만약 객체지향 언어를 구현한다면 반복적으로 사용될 것으로 판단되는 기능(function = 함수) 사전에 만들어놓고
# 필요에 따라 해당 함수를 호출하거나, 또한 반복적으로 사용될 것으로 판단되는 속성(property =속성, 객체의 구성요소)
# 객체(key = value, property)들 사전에 한 번 세팅해놓고 필요에 따라서 가져다가 사용할 수 있도록 함
# 따라서 절차지향 언어보다 훨씬 반복적인 코딩 작업 시, 시간을 효율적으로 사용, 빠르게 코드를 실행
# 파이썬 언어 => 현 시점, 객체지향 대표주자!!!
# 모든 객체지향 프로그래밍 언어들은 반드시 최초에 함수, 속성을 정의할 수 있도록 Class를 선언

```

```python
# 클래스로 탄생된 애들만 갖다 쓸 수 있는 함수, 생성자 함수
class Student : #스튜던트라는 클래스를 선언함
    def __init__(self, name, grade) : #이름 변경 불가, 우리가 바꿀 수 있는 것은 클래스명 뿐이야ㅠㅠ#dlstmxjstmrorcpfmf todtjdgkrl dnlgks ahrwjrdmfh dlstmxjsxmfmf t?
        self.name = name
        self.grade = grade
    def study(self) : # 메서드
        print(self.name, '가(이) 공부합니다.')
# 얘네가 클래스고 얘네가 객체다?!
```

```python
# 독립적인 함수, 아무데나 갖다쓸 수 았음
def
```

```python
student1 = Student("David", 3)
student2 = Student("Jhon", 2)
```

```python
student1.study()
```
