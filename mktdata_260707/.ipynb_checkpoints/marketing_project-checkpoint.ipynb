```python
import pymysql
import requests # modul이라고 되어있음, 객체라는 뜻 객체 뒤에는 메소드가 옴?
from bs4 import BeautifulSoup

db = pymysql.connect(
    host="localhost",
    port=3306,
    user="root",
    passwd="0000",
    db="ecommerce",
    charset="utf8"
)

# 직접 쿼리문을 작성해서 무언가를 실행할 수 있다
cursor = db.cursor()


# res = requests.get("https://davelee-fun.github.io/")

for page_num in range(8):
    if page_num == 0:
        res = requests.get("https://davelee-fun.github.io/")
    else:
        res = requests.get(f"https://davelee-fun.github.io/page{str(page_num+ 1)}")

# 세탁기에 빨래륻 돌렸다! 우리가 보기 편하게 아름답게 정리됐다!
    soup = BeautifulSoup(res.content, "html.parser")

# find와 모가 달름? 리스트 형태로 반환!
    data = soup.select("div.card-body")

    for item in data :
        category = item.select_one("h2.card-title").get_text().replace("관련 상품 추천", "").strip()
        product = item.select_one("h4.card-text").get_text().replace("상품명", "").strip()

        SQL = f"""INSERT INTO teddyproducts(title, category) VALUES('{product}', '{category}')"""
        cursor.execute(SQL)
        
db.commit()
db.close() #메모리 누수 차단
```

```python
import pymysql
import requests # modul이라고 되어있음, 객체라는 뜻 객체 뒤에는 메소드가 옴?
from bs4 import BeautifulSoup

db = pymysql.connect(
    host="localhost",
    port=3306,
    user="root",
    passwd="0000",
    db="ecommerce",
    charset="utf8"
)

# 직접 쿼리문을 작성해서 무언가를 실행할 수 있다
cursor = db.cursor()

SQL = f"""SELECT * FROM teddyproducts WHERE category = '행거도어';"""
cursor.execute(SQL) #임시메모리공간 사용, 실제 저장을 하려면 commit을 써야함

rows = cursor.fetchall() #저..전부 찾아?;;
for row in rows:
    print(row)

db.close()
```
